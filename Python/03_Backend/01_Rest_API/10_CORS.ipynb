# 10 - CORS (Cross-Origin Resource Sharing)

## 1. What is CORS?

**Definition:** A **browser** security rule: a web page from one origin can't read responses from a different origin unless that server allows it.

**Origin = protocol + domain + port**
- http://localhost:3000 and http://localhost:8000 → different origins (port differs)
- https://www.shopkart.com and https://api.shopkart.com → different origins (domain differs)

**How it works:**
1. Browser adds an **Origin** header to the request, e.g. Origin: https://www.shopkart.com
2. Backend compares it with its **allow_origins** list
3. Match → browser lets the page read the response. No match → browser blocks it

**Key points:**
- Only **browsers** enforce CORS. Postman, scripts and server-to-server calls ignore it
- The backend URL in frontend code is **not secret**; anyone can see it in dev tools
- "*" = any website can call your API from a browser. Only for local testing
- In production, list only real frontend domains
- Pods and Docker don't matter; only the domain the user opens in the browser
- **CORS is not authentication**: CORS controls which **site**, JWT controls which **user**
- **Proxy:** if frontend and backend sit behind one domain (www.shopkart.com/api → backend via Nginx), the browser sees one origin and CORS never applies

**Real-world example:** Frontend www.shopkart.com, API api.shopkart.com allowing only the shopkart domain. A fake site's hidden call to /orders from your browser is blocked.

**Interview line:** "CORS is a browser rule. The browser sends the Origin header and the backend allows only listed frontend domains. In production I replace the wildcard with real domains using CORSMiddleware. CORS controls which site can call us, JWT controls which user. We need both."

```python
# ==================================================
# 1. CORS CHECK - what the browser does
# ==================================================
def cors_check(request_origin, allow_origins):
    if "*" in allow_origins or request_origin in allow_origins:
        return "ALLOWED - browser lets the page read the response"
    return "BLOCKED - browser hides the response"

origins_to_test = ["https://www.shopkart.com", "https://free-offers-site.com", "http://localhost:4000"]

print("=" * 65)
print('[CONFIG] allow_origins = ["*"]   (BAD for production)')
for o in origins_to_test:
    print(f"   {o:32} -> {cors_check(o, ['*'])}")

print('[CONFIG] allow_origins = ["https://www.shopkart.com"]   (GOOD)')
for o in origins_to_test:
    print(f"   {o:32} -> {cors_check(o, ['https://www.shopkart.com'])}")
print("=" * 65)

# FastAPI equivalent (for reference):
# app.add_middleware(CORSMiddleware,
#     allow_origins=["https://www.shopkart.com"],
#     allow_methods=["*"], allow_headers=["*"])
```

## Key Takeaways

- Origin = protocol + domain + port
- Only browsers enforce CORS
- "*" = any site can call you; use a list in production
- CORS = which site, JWT = which user; need both
- Same-domain proxy avoids CORS entirely
