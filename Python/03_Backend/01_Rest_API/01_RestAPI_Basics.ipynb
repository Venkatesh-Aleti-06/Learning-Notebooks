# 01 - REST Basics and URL Naming

## 0. History, Constraints, and REST vs RESTful

**History:** In 2000, Roy Fielding, one of the authors of the HTTP specification, defined REST in his PhD thesis. It became the standard for web APIs because it uses plain HTTP, which every system already understands.

**The six REST constraints:**

1. **Client-server:** frontend and backend are separate, so each can change without the other.
   *Example:* the shopping app's UI is redesigned without touching the backend API.
2. **Stateless:** the server doesn't remember earlier requests; every request carries everything it needs.
   *Example:* every call sends the JWT; no session is stored on the server.
3. **Cacheable:** responses say whether they can be cached and for how long.
   *Example:* a product list is cached for 5 minutes, so repeat visits load faster.
4. **Uniform interface:** every resource is handled the same way, with URLs and standard HTTP methods.
   *Example:* GET /orders/5 and GET /users/5 follow the same pattern.
5. **Layered system:** the client doesn't know if it's talking to the real server or a middle layer (load balancer, proxy, cache).
   *Example:* requests pass through Nginx before reaching FastAPI, and the app never notices.
6. **Code on demand (optional):** the server can send code for the client to run.
   *Example:* a server sends JavaScript to the browser. Rarely used in APIs.

**REST vs RESTful:**
- **REST** = the design style (the rules above).
- **RESTful** = an API that actually follows those rules.
- *Example:* GET /orders/5 with proper status codes is RESTful. POST /getOrder?id=5 uses HTTP but breaks the rules, so it isn't RESTful.

**Interview line:** "Roy Fielding defined REST in 2000. It has six constraints, the key ones being client-server, stateless, cacheable and uniform interface. An API that follows them is called RESTful."

## 1. What is REST?

**Definition:** REST (Representational State Transfer) is a style for designing APIs where everything is a **resource** (order, user, product) with its own **URL**, and **HTTP methods** act on it.

**Real-world example:** A shopping site exposes /products, /orders and /users. The app calls GET /orders/5 to see order 5.

**Key points:**
- Resource = a thing (noun), not an action
- URL identifies the resource, HTTP method says what to do
- Data usually travels as JSON
- Stateless: every request carries everything the server needs (for example the JWT)

**Interview line:** "REST models the system as resources with URLs, and uses HTTP methods like GET, POST, PUT, PATCH and DELETE to act on them."

```python
# ==================================================
# 1. WHAT IS REST - a resource returned by GET /orders/5
# ==================================================
order = {"id": 5, "customer": "Aleti", "item": "Laptop", "status": "paid"}

print("=" * 50)
print("[RESOURCE] GET /orders/5")
print("=" * 50)
for key, value in order.items():
    print(f"  {key:10}: {value}")
```

## 2. URL Naming Rules

**Definition:** URLs should name **resources (nouns, plural)**. The HTTP method carries the action.

**Rules:**
- Use nouns, not verbs: /orders, not /getOrders
- Use plurals: /orders, /orders/5
- Lowercase with hyphens: /pbc-reports
- Nest for relationships: /orders/5/items
- Fields go in the **body**, not the URL

**Real-world example (my project):** POST /generate-PBC becomes POST /pbc-reports

**Interview line:** "I name URLs with plural nouns and let the HTTP method express the action, for example POST /pbc-reports instead of /generate-PBC."

```python
# ==================================================
# 2. URL NAMING - BAD vs GOOD
# ==================================================
bad_vs_good = [
    ("GET  /getAllOrders",    "GET    /orders"),
    ("POST /createOrder",     "POST   /orders"),
    ("GET  /order?id=5",      "GET    /orders/5"),
    ("POST /deleteOrder/5",   "DELETE /orders/5"),
    ("GET  /getOrderItems/5", "GET    /orders/5/items"),
    ("POST /generate-PBC",    "POST   /pbc-reports"),
]

print("=" * 55)
print(f"{'BAD':28}GOOD")
print("=" * 55)
for bad, good in bad_vs_good:
    print(f"{bad:28}{good}")
```

## Key Takeaways

- REST = resources + URLs + HTTP methods
- Six constraints; key ones: client-server, stateless, cacheable, uniform interface
- URLs are **plural nouns**, lowercase, hyphenated
- Actions come from the **method**, data goes in the **body**
- RESTful = an API that actually follows REST rules
