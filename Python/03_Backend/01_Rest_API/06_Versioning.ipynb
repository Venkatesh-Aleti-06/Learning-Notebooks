# 06 - API Versioning

## 1. What is Versioning?

**Definition:** Versioning lets you change an API **without breaking existing clients**. Old clients keep using v1 while new clients move to v2.

**When to create a new version:** Only for **breaking changes**:
- Removing or renaming a field
- Changing a field's type (e.g. price from number to string)
- Changing the response shape or behaviour

**Not breaking (no new version needed):** adding a new optional field or a new endpoint.

## 2. Ways to Version

| Method | Example | Notes |
|---|---|---|
| URL path | /v1/orders, /v2/orders | Most common, clear, easy to test |
| Header | Accept-Version: 2 | Cleaner URLs, harder to test in browser |
| Query parameter | /orders?version=2 | Simple but less common |

**Real-world example:** A mobile app on users' phones still calls /v1/orders. The backend launches /v2/orders where "name" is split into "first_name" and "last_name". Old app versions keep working until users update, then v1 is retired.

**Interview line:** "I version in the URL like /v1, create a new version only for breaking changes, and keep the old version running until clients migrate."

```python
# ==================================================
# 1. v1 vs v2 - a breaking change handled safely
# ==================================================
db_user = {"id": 5, "first_name": "Venkatesh", "last_name": "Aleti"}

def get_user_v1(user):
    # old shape: single "name" field (old mobile apps expect this)
    return {"id": user["id"], "name": f"{user['first_name']} {user['last_name']}"}

def get_user_v2(user):
    # new shape: split names (breaking change -> new version)
    return {"id": user["id"], "first_name": user["first_name"], "last_name": user["last_name"]}

routes = {"/v1/users/5": get_user_v1, "/v2/users/5": get_user_v2}

print("=" * 55)
for path, handler in routes.items():
    print(f"[GET {path}] -> {handler(db_user)}")
print("=" * 55)
print("[NOTE] same database row, two response shapes; old clients keep working")
```

## Key Takeaways

- Versioning lets the API change without breaking old clients
- New version only for **breaking** changes (remove, rename, change type)
- Adding optional fields or new endpoints is not breaking
- URL versioning (/v1, /v2) is the most common
- Keep the old version running until clients migrate, then retire it
