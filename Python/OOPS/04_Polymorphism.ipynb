# 04. Polymorphism

## Definition
Polymorphism means **"many forms"** (poly = many, morph = forms). The **same method call** behaves differently depending on the object it's called on.

## Real-world example
A **"pay" button**: the same action works for card, UPI or wallet, but each one processes the payment its own way.

In backend code: you loop over a list of notifiers and call `send()` on each. Email sends an email, SMS sends a text, and your loop never changes.

## Key points
- **Method overriding (runtime polymorphism):** a child redefines a parent method. This is the main kind in Python.
- **Method overloading (compile-time):** the same name with different parameters. Java and C# support it, **Python doesn't**.
- Why Python has no overloading: methods are stored **by name**, so a later definition **replaces** the earlier one.
- Workarounds: **default arguments** or `*args`.
- **Duck typing:** polymorphism **without inheritance**. "If it walks like a duck and quacks like a duck…" If an object has the method, it works.
- **Operator overloading** with dunder methods, like `__add__`, is also a form of polymorphism.

## Interview line
"Polymorphism lets one interface work with many types. In Python that's mainly method overriding and duck typing. Python doesn't support method overloading, so I use default arguments or *args instead."

```python
"""
========================================================
    POLYMORPHISM IN PYTHON - COMPLETE TEACHING DEMO
========================================================
Poly  = many
Morph = forms
Same method call -> different behavior based on object.
========================================================
"""


# ============================================================
# PART 1 : METHOD OVERRIDING  (runtime polymorphism)
# ============================================================
print("############################################################")
print("#   PART 1 : METHOD OVERRIDING (runtime polymorphism)      #")
print("############################################################\n")


class PaymentMethod:
    def pay(self, amount):
        print(f"[PARENT] Generic pay() called for amount {amount}")


class CreditCard(PaymentMethod):
    def pay(self, amount):
        print(f"[CHILD-CARD] Charging {amount} via Credit Card")
        print("[CHILD-CARD] -> contacting Visa/Mastercard network")


class UPI(PaymentMethod):
    def pay(self, amount):
        print(f"[CHILD-UPI] Charging {amount} via UPI")
        print("[CHILD-UPI] -> opening GPay/PhonePe app flow")


class NetBanking(PaymentMethod):
    def pay(self, amount):
        print(f"[CHILD-NB] Charging {amount} via Net Banking")
        print("[CHILD-NB] -> redirecting to bank portal")


print(">>> Creating a list of DIFFERENT payment objects")
payments = [CreditCard(), UPI(), NetBanking()]
print(f"[INFO] payments list length = {len(payments)}\n")

print(">>> Looping and calling the SAME method pay() on each")
print(">>> Notice: we never check the type. Python picks the right one.\n")
for p in payments:
    print(f"[LOOP] object type = {type(p).__name__}")
    p.pay(500)
    print("-" * 50)

print("\n[KEY POINT] Same call `p.pay(500)` -> three different behaviors.")
print("[KEY POINT] This is METHOD OVERRIDING = runtime polymorphism.\n\n")


# ============================================================
# PART 2 : NO METHOD OVERLOADING IN PYTHON
# ============================================================
print("############################################################")
print("#   PART 2 : NO METHOD OVERLOADING IN PYTHON               #")
print("############################################################\n")


class Calculator:
    def add(self, a, b):
        print(f"[VERSION-1] add(a, b) called -> {a} + {b}")
        return a + b

    # This REPLACES the one above. Python keeps only the last definition.
    def add(self, a, b, c):
        print(f"[VERSION-2] add(a, b, c) called -> {a} + {b} + {c}")
        return a + b + c


calc = Calculator()
print(">>> Calling calc.add(10, 20, 30)")
print(f"[RESULT] {calc.add(10, 20, 30)}\n")

print(">>> Now trying calc.add(10, 20) -> this will FAIL")
try:
    calc.add(10, 20)
except TypeError as e:
    print(f"[ERROR] {e}")

print("\n[KEY POINT] Second definition REPLACED the first.")
print("[KEY POINT] Python stores methods by name -> one name, one function.")
print("[KEY POINT] So method overloading does NOT exist in Python.\n\n")


# ============================================================
# PART 3 : THE PYTHONIC FIX -> DEFAULT ARGS / *args
# ============================================================
print("############################################################")
print("#   PART 3 : HOW PYTHON REPLACES OVERLOADING               #")
print("############################################################\n")


class SmartCalculator:
    def add(self, *numbers):
        print(f"[SMART] add() received {len(numbers)} numbers -> {numbers}")
        return sum(numbers)


sc = SmartCalculator()
print(">>> sc.add(10, 20)")
print(f"[RESULT] {sc.add(10, 20)}\n")

print(">>> sc.add(10, 20, 30)")
print(f"[RESULT] {sc.add(10, 20, 30)}\n")

print(">>> sc.add(1, 2, 3, 4, 5)")
print(f"[RESULT] {sc.add(1, 2, 3, 4, 5)}\n")

print("[KEY POINT] One method handles any number of args using *args.\n\n")


# ============================================================
# PART 4 : DUCK TYPING  (Python's special power)
# ============================================================
print("############################################################")
print("#   PART 4 : DUCK TYPING                                   #")
print("############################################################\n")
print("'If it walks like a duck and quacks like a duck, it's a duck.'")
print("No shared parent needed. Just have the right method.\n")


class EmailNotifier:
    def send(self, msg):
        print(f"[EMAIL] Sending email -> {msg}")


class SMSNotifier:
    def send(self, msg):
        print(f"[SMS] Sending SMS -> {msg}")


class SlackNotifier:
    def send(self, msg):
        print(f"[SLACK] Posting to Slack -> {msg}")


# NOTE: no common parent class. No ABC. Nothing.
# They just happen to have send().
notifiers = [EmailNotifier(), SMSNotifier(), SlackNotifier()]

print(">>> Looping over notifiers with NO shared parent class")
for n in notifiers:
    print(f"[LOOP] object type = {type(n).__name__}")
    n.send("Your order is confirmed")
    print("-" * 50)

print("\n[KEY POINT] No inheritance. No interface. It just works.")
print("[KEY POINT] This is DUCK TYPING -> polymorphism without inheritance.\n\n")


# ============================================================
#                      KEY TAKEAWAYS
# ============================================================
print("============================================================")
print("                      KEY TAKEAWAYS                         ")
print("============================================================")
print("1. Polymorphism = one interface, many forms.")
print("2. Overriding (runtime) IS supported in Python. Main kind.")
print("3. Overloading (compile-time) is NOT supported in Python.")
print("   -> second definition replaces the first.")
print("   -> fix with default args or *args.")
print("4. Duck typing: if the method exists, call it. No parent needed.")
print("5. Real backend uses: payments, notifications, storage backends.")
print("============================================================")
```
