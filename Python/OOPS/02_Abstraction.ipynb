# 02. Abstraction

## Definition
Abstraction means **hiding complex implementation details** and showing only the essential features. You know **what** something does, not **how** it does it.

## Real-world example
A **car**: you use the steering wheel, brake and accelerator without knowing how the engine or brakes work inside.

In backend code: a `PaymentGateway` abstract class defines `pay()`. Your service just calls `pay()`, whether it's Razorpay, Stripe or UPI underneath.

## Key points
- Encapsulation hides **data**. Abstraction hides **complexity**.
- In Python, use the `abc` module: inherit from `ABC` and mark methods with `@abstractmethod`.
- You **can't create an object** of an abstract class.
- A child class **must implement every** abstract method, or it can't be instantiated either.
- Python catches a missing method at **runtime**, when you create the object. C# and Java catch it at **compile time**.
- **Duck typing** is a looser alternative: no base class, just matching method names.
- Why interfaces matter: a **contract**, **polymorphism**, and **separation** of what from how.

## Interview line
"Abstraction exposes only what a caller needs and hides the implementation. In Python I use ABC with abstract methods to define a contract that every implementation must follow."

```python
from abc import ABC, abstractmethod

# ============================================================
# ============== THE INTERFACE (ABSTRACT CLASS) ==============
# ============================================================
# PaymentMethod is the CONTRACT.
# It says: "every payment type MUST have a pay() method".
# It does NOT say HOW to pay - that is each child's job.
class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        # No body here on purpose.
        # It is only a PROMISE that this method must exist in every child.
        pass


# ============================================================
# ================ CHILD #1 - CREDIT CARD ====================
# ============================================================
# This child HONOURS the contract -> it implements pay().
class CreditCard(PaymentMethod):
    def pay(self, amount):
        print(f"[CreditCard] Contacting bank...")
        print(f"[CreditCard] Charging {amount} to the card.")
        print(f"[CreditCard] Payment of {amount} successful via Credit Card.")


# ============================================================
# ==================== CHILD #2 - UPI ========================
# ============================================================
# This child also HONOURS the contract -> it implements pay().
class UPI(PaymentMethod):
    def pay(self, amount):
        print(f"[UPI] Opening UPI app...")
        print(f"[UPI] Sending {amount} to merchant.")
        print(f"[UPI] Payment of {amount} successful via UPI.")


# ============================================================
# =========== CHILD #3 - CASH (BROKEN ON PURPOSE) ============
# ============================================================
# This child does NOT implement pay() - it breaks the contract.
# Python will refuse to create this object.
class Cash(PaymentMethod):
    pass   # forgot to implement pay()!


# ============================================================
# ========================== DRIVER ==========================
# ============================================================
print("ABSTRACTION IN PYTHON USING THE ABC MODULE")
print("This program shows how an abstract class enforces a contract.\n")


print("========================================")
print("   STEP 1 : Create the GOOD children")
print("========================================")
print("Creating CreditCard object...")
card = CreditCard()
print("  -> CreditCard object created successfully.")
print("Creating UPI object...")
upi = UPI()
print("  -> UPI object created successfully.\n")


print("========================================")
print("   STEP 2 : Call pay() on each child")
print("========================================")
print(">>> card.pay(1000)")
card.pay(1000)
print()
print(">>> upi.pay(500)")
upi.pay(500)
print()


print("========================================")
print("   STEP 3 : Polymorphism preview")
print("========================================")
print("Both objects go into ONE list.")
print("We loop over them and call pay() the SAME way on each.")
print("The interface GUARANTEES every object has a pay() method,")
print("so we don't need to check the type.\n")

methods = [card, upi]
for m in methods:
    print(f"Calling pay(2000) on a {type(m).__name__} object...")
    m.pay(2000)
    print()


print("========================================")
print("   STEP 4 : Try to create the BROKEN child")
print("========================================")
print("Cash class forgot to implement pay().")
print("Python should REFUSE to create this object.\n")
print(">>> Cash()")
try:
    c = Cash()
    print("  -> Cash object created (this should NOT happen).")
except TypeError as e:
    print(f"[ERROR] Python refused to create the object.")
    print(f"[ERROR] Reason -> {e}")
    print("[ERROR] This is the CONTRACT being enforced at runtime.")
print()


print("========================================")
print("           KEY TAKEAWAY")
print("========================================")
print("- Python has NO 'interface' keyword.")
print("- We use the ABC module: inherit from ABC, mark methods with @abstractmethod.")
print("- An abstract method has NO body - only a promise it must exist.")
print("- Every child MUST implement every abstract method.")
print("- If a child misses one, Python refuses to create the object at RUNTIME.")
print("- (C# / Java catch this at BUILD time - Python catches it at RUNTIME.)")
print("- Why interfaces exist, over a normal child class:")
print("    1. CONTRACT   -> forces consistency across all children.")
print("    2. POLYMORPHISM -> same call works on any child (see STEP 3).")
print("    3. SEPARATION -> interface defines WHAT, each child defines HOW.")
```
