# 02. Open/Closed Principle (OCP)

## Definition
Code should be **open for extension, but closed for modification**. You add new behaviour by adding new code, not by editing working code.

## Real-world example
A payment function with `if method == "card" ... elif method == "upi" ...` has to be edited for every new payment type. Instead, create an abstract `PaymentMethod` with one class per type. Adding `WalletPayment` means writing a new class, and nothing else changes.

## Key points
- Smell: **long if/elif chains on a type**.
- Fix: abstraction plus polymorphism, with one class per variant.
- Fewer edits to tested code means fewer new bugs.

## Interview line
"Open/Closed means I extend behaviour by adding new classes instead of editing existing ones, usually by replacing if/elif chains with an abstract base class."

```python
from abc import ABC, abstractmethod

print("=" * 60)
print("   O - OPEN/CLOSED PRINCIPLE")
print("   Open for extension, closed for modification")
print("=" * 60)

print("\n" + "#" * 60)
print("# BAD: if/elif chain -> must EDIT for every new type")
print("#" * 60)

class PaymentProcessorBad:
    def pay(self, method, amount):
        print(f"[BAD] pay({method}, {amount})")
        if method == "card":
            print("   [BAD] charging card")
        elif method == "upi":
            print("   [BAD] charging UPI")
        else:
            print(f"   [BAD] ERROR: '{method}' not supported -> must edit this class")

bad = PaymentProcessorBad()
bad.pay("card", 500)
bad.pay("wallet", 500)
print("[PROBLEM] New method = edit tested code = risk of breaking card/UPI")

print("\n" + "#" * 60)
print("# GOOD: abstraction + new class per type")
print("#" * 60)

class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CardPayment(PaymentMethod):
    def pay(self, amount):
        print(f"   [CARD] charging {amount}")

class UPIPayment(PaymentMethod):
    def pay(self, amount):
        print(f"   [UPI] charging {amount}")

class PaymentProcessor:
    # NEVER changes again
    def process(self, method: PaymentMethod, amount):
        print(f"[PROCESSOR] processing via {type(method).__name__}")
        method.pay(amount)

processor = PaymentProcessor()
processor.process(CardPayment(), 500)
processor.process(UPIPayment(), 500)

print("\n>>> Business wants WALLET. We ADD a class, edit nothing.")

class WalletPayment(PaymentMethod):
    def pay(self, amount):
        print(f"   [WALLET] charging {amount}")

processor.process(WalletPayment(), 500)
print("[RESULT] PaymentProcessor untouched. Card and UPI untouched.")

print("\n" + "=" * 60)
print("KEY TAKEAWAY")
print("=" * 60)
print("Extend by ADDING new classes, not EDITING old ones.")
print("Tool: abstraction + polymorphism.")
print("Smell: long if/elif chains on type.")
```
