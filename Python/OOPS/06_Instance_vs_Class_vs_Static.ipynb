# 06. Class, Static and Instance Methods

## Definition
Python classes have three kinds of methods, depending on **what they need access to**:
- **Instance method:** takes `self`. Works with **one object's** data.
- **Class method:** takes `cls`, marked `@classmethod`. Works with **class-level** data shared by all objects.
- **Static method:** takes **neither**, marked `@staticmethod`. A utility function grouped inside the class.

## Real-world example
An **Employee** class:
- `get_details()` is an instance method: it reads this employee's name and salary.
- `from_string("Ravi-50000")` is a class method: an alternative constructor, and it updates the shared employee count.
- `is_valid_email(email)` is a static method: it just checks a string and needs no employee data.

## Key points
- **Quick test:** uses `self`? Instance. Uses `cls`? Class. Neither? Static.
- Class methods are the standard way to write **alternative constructors**.
- Static methods are **not** about "fixed values" or saving memory.
- **Java trap:** in Java, `static` also means shared data. In Python, shared data is a **class variable**, not a static method.
- Calling an instance method on the class without an object raises `TypeError`.

## Interview line
"Instance methods work on one object through self, class methods work on class-level state through cls and are used for alternative constructors, and static methods are utilities that need neither."

```python
print("=" * 60)
print("   INSTANCE vs CLASS vs STATIC METHODS - DEMO")
print("=" * 60)

class Employee:
    company = "Accenture"      # class variable -> shared by ALL employees
    count = 0                  # class variable -> total employees created

    def __init__(self, name, salary):
        self.name = name       # instance variable -> unique per object
        self.salary = salary
        Employee.count += 1
        print(f"[INIT] Created {self.name} | total employees = {Employee.count}")

    # ---------------- INSTANCE METHOD ----------------
    # gets self -> ONE specific object
    def show(self):
        print(f"[INSTANCE] self = {self.name} -> salary {self.salary}, company {self.company}")

    # ---------------- CLASS METHOD ----------------
    # gets cls -> the CLASS itself
    @classmethod
    def total(cls):
        print(f"[CLASS] cls = {cls.__name__} -> total employees = {cls.count}")

    @classmethod
    def from_string(cls, text):
        print(f"[CLASS] from_string('{text}') -> splitting and calling cls(...)")
        name, salary = text.split(",")
        return cls(name, int(salary))

    # ---------------- STATIC METHOD ----------------
    # gets NOTHING -> plain utility function
    @staticmethod
    def is_valid_email(email):
        result = "@" in email and "." in email
        print(f"[STATIC] is_valid_email('{email}') -> {result}  (no self, no cls)")
        return result


print("\n" + "#" * 60)
print("# 1. INSTANCE METHOD -> needs ONE object")
print("#" * 60)
ravi = Employee("Ravi", 50000)
sita = Employee("Sita", 70000)
ravi.show()
sita.show()
print("[NOTE] Same method, different output -> each uses its own self")

print("\n" + "#" * 60)
print("# 2. CLASS METHOD -> works on the CLASS")
print("#" * 60)
Employee.total()
print("\n>>> Alternative constructor:")
arjun = Employee.from_string("Arjun,60000")
arjun.show()
Employee.total()

print("\n" + "#" * 60)
print("# 3. STATIC METHOD -> needs NOTHING")
print("#" * 60)
Employee.is_valid_email("ravi@accenture.com")
Employee.is_valid_email("bad-email")
print("[NOTE] Called on the class, no object needed")

print("\n" + "#" * 60)
print("# 4. What happens if instance method is called on class?")
print("#" * 60)
try:
    Employee.show()
except TypeError as e:
    print(f"[ERROR] {e}")
print("[NOTE] Instance method NEEDS an object -> no self, no call")

print("\n" + "=" * 60)
print("KEY TAKEAWAY")
print("=" * 60)
print("Instance -> self -> one object's data      (most common)")
print("Class    -> cls  -> class data / alt constructors")
print("Static   -> none -> utility, logically grouped")
print("Test: uses self? instance. uses cls? class. neither? static.")
```
