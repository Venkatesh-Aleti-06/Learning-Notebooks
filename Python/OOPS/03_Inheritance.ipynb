# 03. Inheritance

## Definition
Inheritance lets a **child class reuse and extend** the attributes and methods of a **parent class**. It models an **"is a"** relationship.

## Real-world example
A **Dog is an Animal**. Every animal can eat and sleep, so the Dog gets those for free, and adds its own `bark()`.

In backend code: a `BaseRepository` holds common CRUD methods, and `UserRepository` and `OrderRepository` inherit them instead of rewriting them.

## Key points
- **Terms:** parent, base or superclass; child, derived or subclass.
- **Overriding:** the child **replaces** a parent method with its own version.
- **Extending with `super()`:** the child calls the parent method, then adds to it.
- Inheritance flows **downward only**. A parent can't access a child's methods.
- **Types:** single, multilevel and multiple. Python allows **multiple** inheritance; Java and C# don't.
- **Diamond problem:** when two parents share a grandparent. Python resolves it with the **MRO** (Method Resolution Order), checked with `ClassName.__mro__`.
- **Prefer composition** ("has a") when the relationship isn't truly "is a".

## Interview line
"Inheritance lets a child class reuse a parent's code and override or extend it with super(). Python supports multiple inheritance and resolves conflicts using the MRO, but I prefer composition when it's not a true is-a relationship."

```python
# =============================================================
# =============================================================
# ========= INHERITANCE IN PYTHON - ALL THREE TYPES ===========
# =============================================================
# =============================================================
# This program shows:
#   1. SINGLE inheritance     (one parent, one child)
#      - Including override and super() extend
#   2. MULTILEVEL inheritance (chain: A -> B -> C)
#   3. MULTIPLE inheritance   (one child, two parents)
#      - Including the Diamond Problem and the MRO
# =============================================================
```

```python
# #############################################################
# ############## PART 1 : SINGLE INHERITANCE ##################
# #############################################################
print("#############################################################")
print("############## PART 1 : SINGLE INHERITANCE ##################")
print("#############################################################\n")

# ===== Parent class =====
class Animal:
    def __init__(self, name):
        print(f"[Animal.__init__] Setting name = {name}")
        self.name = name

    def eat(self):
        print(f"[Animal.eat] {self.name} is eating.")

    def sleep(self):
        print(f"[Animal.sleep] {self.name} is sleeping like a normal animal.")


# ===== Child class =====
# - Inherits eat() AS-IS from Animal
# - OVERRIDES sleep() completely (parent's version is thrown away)
# - EXTENDS __init__ using super() (parent's logic + extra)
class Dog(Animal):
    def __init__(self, name, breed):
        print(f"[Dog.__init__] Called with name={name}, breed={breed}")
        print("[Dog.__init__] Calling super().__init__() to run parent setup first...")
        super().__init__(name)                     # EXTEND the parent constructor
        print(f"[Dog.__init__] Now adding my own field: breed = {breed}")
        self.breed = breed

    def sleep(self):                               # OVERRIDE
        print(f"[Dog.sleep] {self.name} the {self.breed} is sleeping curled up.")

    def bark(self):                                # Child-only method
        print(f"[Dog.bark] {self.name} says: Woof!")


print("----- STEP 1 : Create a Dog object -----")
d = Dog("Buddy", "Labrador")
print()

print("----- STEP 2 : Child uses INHERITED parent method as-is -----")
d.eat()                   # inherited from Animal, not touched
print()

print("----- STEP 3 : Child uses its OVERRIDDEN version -----")
d.sleep()                 # Dog's sleep wins, Animal's is thrown away
print()

print("----- STEP 4 : Child calls its OWN method -----")
d.bark()
print()

print("----- STEP 5 : Direction check - parent CANNOT see child's methods -----")
a = Animal("GenericAnimal")
a.eat()
print("Trying a.bark() ...")
try:
    a.bark()
except AttributeError as e:
    print(f"[ERROR] Parent has no access to child method -> {e}")
print()
```

```python
# #############################################################
# ############ PART 2 : MULTILEVEL INHERITANCE ################
# #############################################################
print("\n#############################################################")
print("############ PART 2 : MULTILEVEL INHERITANCE ################")
print("#############################################################\n")
# Chain: Grandfather -> Father -> Son
# Each level adds something, and Son inherits EVERYTHING up the chain.

class Grandfather:
    def house(self):
        print("[Grandfather.house] Owns the family house.")

class Father(Grandfather):                        # Father inherits from Grandfather
    def car(self):
        print("[Father.car] Owns a car.")

class Son(Father):                                # Son inherits from Father (and transitively Grandfather)
    def phone(self):
        print("[Son.phone] Owns a smartphone.")


print("----- Create a Son object -----")
s = Son()
print()

print("----- Son can access methods from ALL levels above -----")
s.house()     # from Grandfather - inherited through Father
s.car()       # from Father
s.phone()     # own method
print()

print("----- But Grandfather cannot access Son's methods -----")
g = Grandfather()
g.house()
print("Trying g.phone() ...")
try:
    g.phone()
except AttributeError as e:
    print(f"[ERROR] Grandfather has no access to Son's method -> {e}")
print()
print(">>> Key point: inheritance flows DOWNWARD, not upward.")
```

```python
# #############################################################
# ############ PART 3 : MULTIPLE INHERITANCE ##################
# ############ (Diamond Problem + MRO)         ################
# #############################################################
print("\n#############################################################")
print("############ PART 3 : MULTIPLE INHERITANCE ##################")
print("############    Diamond Problem + MRO        ################")
print("#############################################################\n")

# The diamond shape:
#
#            A
#           / \
#          B   C
#           \ /
#            D
#
# Both B and C override greet().
# When D calls greet(), WHICH ONE WINS?
# Python answers this using the MRO (Method Resolution Order).

class A:
    def greet(self):
        print("[A.greet] Hello from A.")

class B(A):
    def greet(self):
        print("[B.greet] Hello from B.")

class C(A):
    def greet(self):
        print("[C.greet] Hello from C.")

class D(B, C):                                    # D inherits from BOTH B and C
    pass


print("----- Create a D object -----")
d_obj = D()
print()

print("----- Call greet() on D - which parent wins? -----")
d_obj.greet()
print()

print("----- Why? Check the MRO (Method Resolution Order) -----")
print("Python searches classes in this fixed order to find a method:")
for cls in D.__mro__:
    print(f"   -> {cls.__name__}")
print()
print(">>> Python picked B.greet() because B comes BEFORE C in the MRO.")
print(">>> The order follows class D(B, C) - left to right - then up to A, then object.")
print(">>> This is called C3 linearization. You don't need the algorithm,")
print(">>> just know: MRO decides the winner, searched left-to-right.")


# =============================================================
# ===================== KEY TAKEAWAY ==========================
# =============================================================
print("\n=============================================================")
print("=====================  KEY TAKEAWAY  ========================")
print("=============================================================")
print("- Inheritance = child reuses parent's variables and methods.")
print("- Reason: CODE REUSE. Write common logic once in the parent.")
print()
print("- OVERRIDE  -> child replaces parent's method completely.")
print("- SUPER()   -> child extends parent's method (parent + extra).")
print()
print("- DIRECTION -> child can access parent; parent CANNOT access child.")
print()
print("- TYPES:")
print("    SINGLE     -> one parent, one child.")
print("    MULTILEVEL -> chain: A -> B -> C.")
print("    MULTIPLE   -> one child, two parents (Python allows, Java/C# do not).")
print()
print("- DIAMOND PROBLEM -> conflict in multiple inheritance when two")
print("  parents have the same method.")
print("- MRO (Method Resolution Order) -> Python's fixed search order")
print("  that resolves the conflict. Left-to-right, then upward.")
print("  Check it anytime with ClassName.__mro__")
```
