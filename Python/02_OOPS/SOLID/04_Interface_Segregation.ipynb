# 04. Interface Segregation Principle (ISP)

## Definition
**Don't force a class to implement methods it doesn't need.** Several small, focused interfaces are better than one big one.

## Real-world example
A fat `Notifier` with `send_text()`, `send_attachment()` and `send_html()` forces an SMS class to fake two methods it can't support.
Fix: split it into `TextSender`, `AttachmentSender` and `HtmlSender`. Email implements all three, and SMS implements only `TextSender`.

## Key points
- Smell: classes with **empty or "not supported" methods**.
- Small interfaces mean classes only depend on what they actually use.
- Closely related to Liskov: ISP stops the promise from being **too big**.

## Interview line
"Interface Segregation means I keep interfaces small, so no class has to implement methods it doesn't use."

```python
from abc import ABC, abstractmethod

print("=" * 60)
print("   I - INTERFACE SEGREGATION PRINCIPLE")
print("   Don't force a class to implement methods it doesn't need")
print("=" * 60)

print("\n" + "#" * 60)
print("# BAD: one FAT interface for every notifier")
print("#" * 60)

class NotifierBad(ABC):
    @abstractmethod
    def send_text(self, msg): pass
    @abstractmethod
    def send_attachment(self, file): pass
    @abstractmethod
    def send_html(self, html): pass

class EmailBad(NotifierBad):
    def send_text(self, msg):        print(f"   [EMAIL] text: {msg}")
    def send_attachment(self, file): print(f"   [EMAIL] attachment: {file}")
    def send_html(self, html):       print(f"   [EMAIL] html: {html}")

class SMSBad(NotifierBad):
    def send_text(self, msg):        print(f"   [SMS] text: {msg}")
    def send_attachment(self, file): raise Exception("SMS can't send files!")
    def send_html(self, html):       raise Exception("SMS can't send HTML!")

sms = SMSBad()
sms.send_text("OTP is 1234")
try:
    sms.send_attachment("invoice.pdf")
except Exception as e:
    print(f"   [CRASH] {e}")
print("[PROBLEM] SMS forced to write 2 fake methods it can't support.")
print("[PROBLEM] This also breaks Liskov -> ISP and LSP are related.")

print("\n" + "#" * 60)
print("# GOOD: small, focused interfaces")
print("#" * 60)

class TextSender(ABC):
    @abstractmethod
    def send_text(self, msg): pass

class AttachmentSender(ABC):
    @abstractmethod
    def send_attachment(self, file): pass

class HtmlSender(ABC):
    @abstractmethod
    def send_html(self, html): pass

class Email(TextSender, AttachmentSender, HtmlSender):
    def send_text(self, msg):        print(f"   [EMAIL] text: {msg}")
    def send_attachment(self, file): print(f"   [EMAIL] attachment: {file}")
    def send_html(self, html):       print(f"   [EMAIL] html: {html}")

class SMS(TextSender):
    def send_text(self, msg):        print(f"   [SMS] text: {msg}")

print(">>> Everyone who can send text:")
for n in [Email(), SMS()]:
    n.send_text("Order confirmed")

print("\n>>> Only attachment senders:")
Email().send_attachment("invoice.pdf")

print("[RESULT] SMS implements ONLY what it supports. No fake methods.")

print("\n" + "=" * 60)
print("KEY TAKEAWAY")
print("=" * 60)
print("Many small interfaces > one fat interface.")
print("A class picks only the interfaces it can truly support.")
print("Smell: methods that raise errors or just 'pass'.")
```
