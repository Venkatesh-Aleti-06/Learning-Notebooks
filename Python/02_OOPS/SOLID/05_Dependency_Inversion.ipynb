# 05. Dependency Inversion Principle (DIP)

## Definition
**Depend on abstractions, not concrete classes.** High-level code shouldn't create its own dependencies. They should be **handed in** from outside.

## Real-world example
A **Car** that builds its own `PetrolEngine` inside can never use an electric one. Instead, the Car accepts any `Engine` in its constructor, so you can pass in petrol or electric.
"Don't make your own parts. Get them handed to you."

In backend code: `OrderService` receives a repository instead of creating a Postgres connection itself. In tests, you pass in a fake repository.

## Key points
- **Inversion** is the principle. **Dependency injection** is how you do it, usually through the constructor.
- Makes code **testable** (pass in fakes) and **swappable** (change the database or provider).
- FastAPI's `Depends(get_db)` is dependency injection in action.

## Interview line
"Dependency Inversion means my classes depend on abstractions and receive their dependencies through the constructor, so I can swap implementations or use fakes in tests."

```python
from abc import ABC, abstractmethod

print("=" * 50)
print("  D - DEPENDENCY INVERSION: CAR AND ENGINE")
print("=" * 50)

# ---------------- BAD ----------------
print("\n##### BAD: Car builds its own PetrolEngine #####")

class PetrolEngine:
    def start(self):
        print("   [PETROL] vroom")

class CarBad:
    def __init__(self):
        self.engine = PetrolEngine()      # fixed forever
        print("[BAD] Car made its own PetrolEngine")

    def drive(self):
        self.engine.start()

CarBad().drive()
print("[PROBLEM] Want electric? Must rebuild the Car class.")

# ---------------- GOOD ----------------
print("\n##### GOOD: Car accepts ANY Engine #####")

class Engine(ABC):                        # the rule: every engine can start()
    @abstractmethod
    def start(self): pass

class Petrol(Engine):
    def start(self):
        print("   [PETROL] vroom")

class Electric(Engine):
    def start(self):
        print("   [ELECTRIC] silent hum")

class Car:
    def __init__(self, engine: Engine):   # engine given from outside
        self.engine = engine
        print(f"[CAR] fitted with {type(engine).__name__}")

    def drive(self):
        self.engine.start()

Car(Petrol()).drive()
Car(Electric()).drive()
print("[RESULT] Same Car class. Only the engine changed.")

print("\n" + "=" * 50)
print("KEY TAKEAWAY")
print("=" * 50)
print("Bad : class MAKES its own part  -> stuck with it")
print("Good: class is GIVEN the part   -> swap anytime")
```
