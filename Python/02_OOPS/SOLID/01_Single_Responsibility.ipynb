# 01. Single Responsibility Principle (SRP)

## Definition
A class should have **only one reason to change**, so it does one job only.

## Real-world example
An **OrderManager** that validates orders, saves them to the database **and** sends emails has three reasons to change. Split it into `OrderValidator`, `OrderRepository` and `EmailService`, with an `OrderService` coordinating them.

## Key points
- Smell: a **"God class"** that does everything.
- Smaller classes are easier to **test**, **reuse** and **change**.
- In backend code: route, service and repository layers are SRP applied to the whole app.

## Interview line
"SRP means a class has one reason to change. I split validation, data access and notifications into separate classes, so changing one doesn't break the others."

```python
print("=" * 60)
print("   S - SINGLE RESPONSIBILITY PRINCIPLE")
print("   A class should have only ONE reason to change")
print("=" * 60)

print("\n" + "#" * 60)
print("# BAD: GOD CLASS -> does everything")
print("#" * 60)

class OrderManagerBad:
    def place_order(self, item, price):
        print(f"[BAD] validating order for {item}")
        if price <= 0:
            print("[BAD] invalid price")
            return
        print(f"[BAD] saving {item} to database")
        print(f"[BAD] sending email for {item}")
        print("[BAD] one class = validation + DB + email")

OrderManagerBad().place_order("Laptop", 50000)
print("[PROBLEM] Change email provider? Edit this class.")
print("[PROBLEM] Change database? Edit this class again.")
print("[PROBLEM] 3 reasons to change -> breaks SRP")

print("\n" + "#" * 60)
print("# GOOD: one class, one job")
print("#" * 60)

class OrderValidator:
    def validate(self, price):
        print(f"   [VALIDATOR] checking price {price}")
        return price > 0

class OrderRepository:
    def save(self, item):
        print(f"   [REPOSITORY] saving {item} to database")

class EmailService:
    def send(self, item):
        print(f"   [EMAIL] sending confirmation for {item}")

class OrderService:
    # composition + dependency injection
    def __init__(self, validator, repo, email):
        self.validator = validator
        self.repo = repo
        self.email = email
        print("[SERVICE] created with injected validator, repo, email")

    def place_order(self, item, price):
        print(f"\n[SERVICE] place_order({item}, {price}) -> coordinating")
        if not self.validator.validate(price):
            print("   [SERVICE] rejected")
            return
        self.repo.save(item)
        self.email.send(item)
        print("   [SERVICE] done")

service = OrderService(OrderValidator(), OrderRepository(), EmailService())
service.place_order("Laptop", 50000)
service.place_order("Phone", -10)

print("\n" + "=" * 60)
print("KEY TAKEAWAY")
print("=" * 60)
print("One class = one job = one reason to change.")
print("Change email provider -> only EmailService changes.")
print("Service just coordinates. Easy to test each part alone.")
```
