# 07. Dunder Methods

## Definition
**Dunder** means "double underscore", as in `__init__`. They're special methods Python calls **automatically** for built-in actions like printing, comparing or adding. Every class gets them from the base `object` class, and you override them to customise behaviour.

## Real-world example
A **Book** class: by default, `print(book)` shows a memory address, and two books with the same title aren't equal. With `__str__` and `__eq__`, printing shows the title, and books compare by their data.

## Key points
- `__init__`: runs when an object is created.
- `__str__`: readable text for **users**, used by `print()`.
- `__repr__`: detailed text for **developers**, used in lists, logs and notebooks.
- If `__str__` is missing, Python falls back to `__repr__`, so if you write only one, write `__repr__`.
- `__eq__`: by default `==` compares memory, so override it to compare data.
- `__len__`, `__add__`, `__iter__` and `__next__` make objects work with `len()`, `+` and loops.

## Interview line
"Dunder methods let my classes work with Python's built-in operations. I usually define __repr__ for debugging and __eq__ so objects compare by value instead of identity."

```python
print("=" * 60)
print("PART 1: WITHOUT DUNDER METHODS")
print("=" * 60)

class PlainBook:
    """A class with no dunder methods: Python uses the defaults."""
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages
        print(f"   [INIT] PlainBook created: {title}")

p1 = PlainBook("Python", 300)
p2 = PlainBook("Python", 300)

print(f"[PRINT] print(p1) -> {p1}")
print("        (shows a memory address, not useful)")
print(f"[EQUAL] p1 == p2  -> {p1 == p2}")
print("        (compares memory, not data)")

print("\n" + "=" * 60)
print("PART 2: WITH DUNDER METHODS")
print("=" * 60)

class Book:
    """A class that customises built-in behaviour with dunders."""
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages
        print(f"   [INIT] Book created: {title}")

    def __str__(self):
        """Readable text for users, used by print()."""
        return f"Book: {self.title}"

    def __repr__(self):
        """Detailed text for developers, used in lists and logs."""
        return f"Book(title='{self.title}', pages={self.pages})"

    def __eq__(self, other):
        """Compare books by their data instead of memory."""
        return self.title == other.title and self.pages == other.pages

    def __len__(self):
        """Lets len(book) return the page count."""
        return self.pages

    def __add__(self, other):
        """Lets book1 + book2 return total pages."""
        return self.pages + other.pages

b1 = Book("Python", 300)
b2 = Book("Python", 300)

print(f"\n[STR]   print(b1)  -> {b1}")
print(f"[REPR]  [b1]       -> {[b1]}")
print(f"[EQUAL] b1 == b2   -> {b1 == b2}")
print(f"[LEN]   len(b1)    -> {len(b1)}")
print(f"[ADD]   b1 + b2    -> {b1 + b2}")

print("\n" + "=" * 60)
print("KEY TAKEAWAY")
print("=" * 60)
print("1. __str__ is for users, __repr__ is for developers.")
print("2. If you write only one, write __repr__.")
print("3. Default == compares memory; override __eq__ to compare data.")
print("4. __len__ and __add__ make objects work with len() and +.")
```
