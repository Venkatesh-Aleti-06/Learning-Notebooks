# 08. @property

## Definition
`@property` turns a method into something you **use like an attribute**. You write `account.balance`, not `account.balance()`, while Python runs a method behind the scenes. It's Python's version of getters and setters in Java, or properties in C#.

## Real-world example
A **BankAccount**: reading `account.balance` just returns the value, but setting `account.balance = -500` runs validation and rejects it.

## Key points
- `@property` makes the **getter**.
- `@name.setter` makes the **setter**, and validation goes here.
- The name before `.setter` **must match** the property name.
- **No setter** means the property is **read-only**.
- Useful for **computed values**, like `circle.area` worked out from the radius.
- It's **not** a security feature, just controlled access.

## Interview line
"@property lets me add validation or computed values behind normal attribute access, so the class's public interface stays clean."

```python
print("=" * 60)
print("PART 1: GETTER + SETTER WITH VALIDATION")
print("=" * 60)

class BankAccount:
    """Balance is read and written through a property."""
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance
        print(f"   [INIT] {owner}'s account, balance {balance}")

    @property
    def balance(self):
        """Getter: runs on account.balance"""
        print("   [GETTER] reading balance")
        return self._balance

    @balance.setter
    def balance(self, value):
        """Setter: runs on account.balance = value"""
        print(f"   [SETTER] trying to set {value}")
        if value < 0:
            print("   [REJECTED] balance can't be negative")
            return
        self._balance = value
        print("   [OK] balance updated")

acc = BankAccount("Ravi", 1000)
print(f"[READ] balance = {acc.balance}")
acc.balance = 2000
acc.balance = -500
print(f"[READ] balance = {acc.balance}")

print("\n" + "=" * 60)
print("PART 2: READ-ONLY COMPUTED PROPERTY")
print("=" * 60)

class Circle:
    """area is computed and has no setter, so it's read-only."""
    def __init__(self, radius):
        self.radius = radius

    @property
    def area(self):
        return round(3.14159 * self.radius ** 2, 2)

c = Circle(5)
print(f"[COMPUTED] area = {c.area}")
c.radius = 10
print(f"[COMPUTED] after radius change, area = {c.area}")

try:
    c.area = 100
except AttributeError as e:
    print(f"[READ-ONLY] {e}")

print("\n" + "=" * 60)
print("KEY TAKEAWAY")
print("=" * 60)
print("1. @property = getter, @name.setter = setter.")
print("2. Put validation in the setter.")
print("3. No setter means read-only.")
print("4. Use it like an attribute, no brackets.")
```
