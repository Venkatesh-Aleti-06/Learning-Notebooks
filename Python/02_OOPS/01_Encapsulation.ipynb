# 01. Encapsulation

## Definition
Encapsulation means **bundling data and the methods that work on it inside one class**, and **restricting direct access** to that data. Outside code can only read or change it through controlled methods.

## Real-world example
A **bank account**: you can't reach in and change your balance directly. You have to go through `deposit()` and `withdraw()`, which check the rules, like no negative amounts and no withdrawing more than you have.

In backend code, it's the same idea: a `User` object validates its email inside a setter instead of letting anyone assign a bad value.

## Key points
- **`_name`:** private by **convention**. Python doesn't stop access, but it signals "internal, don't touch".
- **`__name`:** triggers **name mangling**. It becomes `_ClassName__name`, which is harder to reach by accident but still reachable.
- Python has **no true private**, unlike Java or C#.
- It's for **data integrity**, not security.
- **Pattern:** private field + public methods + validation inside them.
- Private **methods** work the same way: `_helper()` is preferred in real code.

## Interview line
"Encapsulation hides an object's internal state and only allows changes through controlled methods, so the data always stays valid."

```python
class BankAccount:

    def __init__(self, owner, initial_balance, pin):
        print("========================================")
        print("During Initialization - Constructor")
        print("========================================")
        # Public attribute - anyone can access this directly
        self.owner = owner
        print(f"Account Owner (public)        : {self.owner}")

        # Single underscore - CONVENTION only (internal, please don't touch)
        self._balance = initial_balance
        print(f"Account Balance (_convention) : {self._balance}")

        # Double underscore - NAME MANGLING (harder to touch from outside)
        self.__pin = pin
        print(f"Account PIN (__mangled)       : {self.__pin}")
        print("[INIT] Account created successfully.\n")

    def deposit(self, amount):
        print("========================================")
        print("             DEPOSIT MONEY")
        print("========================================")
        print(f"Requested Deposit Amount : {amount}")
        print(f"Balance Before Deposit   : {self._balance}")

        # Encapsulation rule: validate before changing the data
        if amount <= 0:
            print("[ERROR] Deposit amount must be positive. Operation blocked.\n")
            return

        self._balance = self._balance + amount
        print(f"[DEPOSIT] Deposited {amount} successfully.")
        print(f"Balance After Deposit    : {self._balance}\n")

    def withdraw(self, amount, pin):
        print("========================================")
        print("            WITHDRAW MONEY")
        print("========================================")
        print(f"Requested Withdraw Amount : {amount}")
        print(f"Balance Before Withdraw   : {self._balance}")

        # Encapsulation rule 1: verify PIN before anything else
        if pin != self.__pin:
            print("[ERROR] Incorrect PIN. Operation blocked.\n")
            return

        # Encapsulation rule 2: never allow overdraw
        if amount > self._balance:
            print("[ERROR] Insufficient balance. Operation blocked.\n")
            return

        # Encapsulation rule 3: must be a positive amount
        if amount <= 0:
            print("[ERROR] Withdraw amount must be positive. Operation blocked.\n")
            return

        self._balance = self._balance - amount
        print(f"[WITHDRAW] Withdrew {amount} successfully.")
        print(f"Balance After Withdraw    : {self._balance}\n")

    def display(self):
              print("===============ACCOUNT SUMMARY===============")
              print(f"Owner   : {self.owner}")
              print(f"Balance : {self._balance}")
              # Notice: we DO NOT expose the PIN here. That is encapsulation.
              print("=============================================\n")
```

```python
print("BANK ACCOUNT PROGRAM")
print("This program demonstrates ENCAPSULATION using Object-Oriented Programming (OOP).\n")

# Create an account object
account = BankAccount(owner="Aleti", initial_balance=43000, pin=1234)

# ===== Normal usage through the methods (the RIGHT way) =====
# account.display()
# account.deposit(27000)
# account.display()
# account.withdraw(8500, pin=1234)
# account.display()

# # ===== Try to break the rules through the methods =====
# print("--- Trying operations that violate the rules ---\n")
# account.deposit(-100)              # negative deposit - blocked
# account.withdraw(99999, "1234")    # over-withdraw - blocked
# account.withdraw(100, "0000")      # wrong PIN - blocked

# # ===== Now show what encapsulation looks like from OUTSIDE the class =====
# print("--- Accessing attributes directly from outside the class ---")

# # Public - works, as expected
print(f"Public (account.owner)         : {account.owner}")

# # Single underscore - STILL WORKS. This is the CONVENTION point.
# # Python did NOT stop us. Only a polite 'please don't touch' sign.
print(f"Convention (account._balance)  : {account._balance}")

# # Double underscore - direct access FAILS (name mangling renamed it)
# try:
#     print(account.__pin)
# except AttributeError as e:
#     print(f"Mangled direct access FAILED  : {e}")

# # ...but you CAN still reach it via the mangled name - so not truly private
print(f"Mangled via real name          : {account._BankAccount__pin}")

# print("\n========================================")
# print("          PROGRAM COMPLETED")
# print("========================================")
# print("\nKEY TAKEAWAY:")
# print("- Public attributes  : anyone can access.")
# print("- Single underscore  : convention, Python does NOT enforce it.")
# print("- Double underscore  : name mangling, harder but NOT truly private.")
# print("- Real protection comes from the methods (deposit/withdraw),")
# print("  which enforce the rules before the data changes.")
```

```python
class BankAccount:

    # ===== The normal constructor =====
    def __init__(self, owner, balance=0, pin=None):
        print("----------------------------------------")
        print("[__init__] Constructor called")
        print("----------------------------------------")
        print(f"Received owner   : {owner}")
        print(f"Received balance : {balance}")
        print(f"Received pin     : {pin}")

        # Assign each field to the object
        self.owner = owner
        print(f"Assigned self.owner   -> {self.owner}")
        self.balance = balance
        print(f"Assigned self.balance -> {self.balance}")
        self.pin = pin
        print(f"Assigned self.pin     -> {self.pin}")
        print("[__init__] Object created successfully.\n")

    # ===== Alternative constructor #1 - build from a dictionary =====
    @classmethod
    def from_dict(cls, data):
        print("----------------------------------------")
        print("[from_dict] Alternative constructor called")
        print("----------------------------------------")
        print(f"Received dictionary : {data}")
        print("Unpacking values from the dictionary...")
        owner   = data["owner"]
        balance = data["balance"]
        pin     = data["pin"]
        print(f"  -> owner   = {owner}")
        print(f"  -> balance = {balance}")
        print(f"  -> pin     = {pin}")
        print("Calling cls(...) which internally triggers __init__ ...")
        return cls(owner, balance, pin)

    # ===== Alternative constructor #2 - new account always starts at zero =====
    @classmethod
    def new_customer(cls, owner, pin):
        print("----------------------------------------")
        print("[new_customer] Alternative constructor called")
        print("----------------------------------------")
        print(f"Received owner : {owner}")
        print(f"Received pin   : {pin}")
        print("Rule for new customer: balance ALWAYS starts at 0.")
        print("Calling cls(...) which internally triggers __init__ ...")
        return cls(owner, 0, pin)


# ============================================================
# ============ DRIVER CODE - THREE WAYS TO CREATE ============
# ============================================================
print("MULTIPLE CONSTRUCTORS IN PYTHON")
print("This program shows the 3 ways to build the same class.\n")


print("========================================")
print("     WAY 1 : Normal Constructor")
print("========================================")
a1 = BankAccount("Aleti", 1000, "1234")
print(f"Final Object -> owner={a1.owner}, balance={a1.balance}, pin={a1.pin}\n")


print("========================================")
print("     WAY 2 : from_dict (classmethod)")
print("========================================")
data = {"owner": "Ravi", "balance": 5000, "pin": "9999"}
a2 = BankAccount.from_dict(data)
print(f"Final Object -> owner={a2.owner}, balance={a2.balance}, pin={a2.pin}\n")


print("========================================")
print("     WAY 3 : new_customer (classmethod)")
print("========================================")
a3 = BankAccount.new_customer("Priya", "0000")
print(f"Final Object -> owner={a3.owner}, balance={a3.balance}, pin={a3.pin}\n")


print("========================================")
print("           KEY TAKEAWAY")
print("========================================")
print("- Python does NOT support constructor overloading.")
print("- There is only ONE real __init__.")
print("- Alternative constructors are built with @classmethod.")
print("- A classmethod prepares arguments its own way, then calls cls(...).")
print("- cls(...) internally triggers __init__ (see the prints above).")
print("- from_dict is a very common pattern in backend work (API/JSON -> object).")
```
