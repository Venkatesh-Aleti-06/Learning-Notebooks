# 05. Multiple Constructors
Python does **not** support constructor overloading. Methods are stored by name, so a second `__init__` **replaces** the first.

**Two ways to get the same effect:**
- **Default parameter values** in a single `__init__`.
- **`@classmethod` factory methods:** they receive `cls` and return `cls(...)`, which runs `__init__`.

**Real use:** `datetime.fromtimestamp()`, `dict.fromkeys()`, or building objects from JSON or DB rows.

```python
print("=" * 55)
print("BAD: trying to overload __init__")
print("=" * 55)

class BadEmployee:
    def __init__(self, name):
        self.name = name
    def __init__(self, name, salary):
        self.name, self.salary = name, salary

try:
    BadEmployee("Ravi")
except TypeError as e:
    print(f"[ERROR] {e}")
print("[NOTE] the second __init__ replaced the first")

print("\n" + "=" * 55)
print("GOOD: defaults + @classmethod factories")
print("=" * 55)

class Employee:
    def __init__(self, name, salary=30000):
        self.name, self.salary = name, salary
        print(f"   [INIT] {name}, salary {salary}")

    @classmethod
    def from_string(cls, data):
        print(f"[FACTORY] from_string('{data}')")
        name, salary = data.split("-")
        return cls(name, int(salary))

    @classmethod
    def intern(cls, name):
        print(f"[FACTORY] intern('{name}')")
        return cls(name, 10000)

print("[DEFAULT] Employee('Ravi')")
Employee("Ravi")
print("[CUSTOM]  Employee('Sita', 50000)")
Employee("Sita", 50000)
Employee.from_string("Anu-60000")
Employee.intern("Kiran")

print("\n" + "=" * 55)
print("KEY TAKEAWAY")
print("=" * 55)
print("No constructor overloading in Python.")
print("Use default args, or @classmethod factories that return cls(...).")
```
