## 1. Variables and Data Types
A variable is a name that points to a value in memory. Python is **dynamically typed**: you don't declare the type, Python works it out from the value.

```python
print("=" * 60)
print("SECTION 1: VARIABLES AND DATA TYPES")
print("=" * 60)

name = "Venkatesh"      # str   - text
age = 25                # int   - whole number
salary = 475000.50       # float - decimal number
is_employed = True      # bool  - True / False
nothing = None          # NoneType - "no value yet"

for label, value in [("name", name), ("age", age), ("salary", salary),
                     ("is_employed", is_employed), ("nothing", nothing)]:
    print(f"[VAR] {label:<12} = {str(value):<10} -> type: {type(value).__name__}")

# Multiple assignment and swapping (no temp variable needed)
a, b = 10, 20
print(f"\n[SWAP] before: a={a}, b={b}")
a, b = b, a
print(f"[SWAP] after : a={a}, b={b}")
```

## 2. Type Conversion (Casting)
Convert between types with `int()`, `float()`, `str()` and `bool()`.

**Truthy and falsy:** values like `0`, `""`, `[]`, `{}` and `None` count as `False` in an if-check. Everything else counts as `True`.

```python
print("=" * 60)
print("SECTION 2: TYPE CONVERSION")
print("=" * 60)

print(f"[CAST] int('42')      -> {int('42')}")
print(f"[CAST] float('3.14')  -> {float('3.14')}")
print(f"[CAST] str(100)       -> '{str(100)}'")
print(f"[CAST] int(9.99)      -> {int(9.99)}   (truncates, does not round)")
print(f"[CAST] round(9.99)    -> {round(9.99)}")

# Truthy / falsy: these values count as False in an if-check
falsy = [0, 0.0, "", [], {}, None, False]
print("\n[TRUTHY] falsy values:", [bool(v) for v in falsy])
print("[TRUTHY] bool('hi') ->", bool("hi"), "| bool([1]) ->", bool([1]))
```

## 3. Operators
- **Arithmetic:** `+  -  *  /  //  %  **`
- `/` always returns a float. `//` is floor division. `%` gives the remainder.
- **Comparison:** `==  !=  >  <  >=  <=`
- **Logical:** `and  or  not`
- **Identity vs equality:** `==` compares values, `is` compares memory identity (common interview trap).
- **Membership:** `in` checks if a value exists in a sequence.

```python
print("=" * 60)
print("SECTION 3: OPERATORS")
print("=" * 60)

x, y = 17, 5

# Arithmetic
print(f"[ARITH] {x} + {y}  = {x + y}")
print(f"[ARITH] {x} - {y}  = {x - y}")
print(f"[ARITH] {x} * {y}  = {x * y}")
print(f"[ARITH] {x} / {y}  = {x / y}")
print(f"[ARITH] {x} // {y} = {x // y}")
print(f"[ARITH] {x} % {y}  = {x % y}")
print(f"[ARITH] {x} ** 2 = {x ** 2}")

# Comparison
print(f"\n[COMPARE] {x} > {y}  -> {x > y}")
print(f"[COMPARE] {x} == {y} -> {x == y}")

# Logical
print(f"\n[LOGIC] True and False -> {True and False}")
print(f"[LOGIC] True or False  -> {True or False}")
print(f"[LOGIC] not True       -> {not True}")

# Identity vs equality
list1 = [1, 2, 3]
list2 = [1, 2, 3]
print(f"\n[EQUAL]    list1 == list2 -> {list1 == list2}")
print(f"[IDENTITY] list1 is list2 -> {list1 is list2}")

# Membership
print(f"\n[IN] 2 in {list1} -> {2 in list1}")
```

## 4. Strings and f-string Formatting
Strings are **immutable**: string methods never change the original, they return a **new** string.

- `.strip()` removes spaces at both ends
- `.upper()` / `.lower()` change case
- `.replace(old, new)` swaps text
- `.split()` breaks a string into a list
- **Slicing:** `s[start:end]`, and `s[::-1]` reverses a string
- **f-strings:** `f"{value}"`, with format options like `{x:,.2f}` for commas and 2 decimals, or `{x:>5}` for right-aligning

```python
print("=" * 60)
print("SECTION 4: STRINGS AND FORMATTING")
print("=" * 60)

s = "  Hello Python  "
print(f"[STR] original   -> '{s}'")
print(f"[STR] .strip()   -> '{s.strip()}'")
print(f"[STR] .upper()   -> '{s.upper()}'")
print(f"[STR] .replace() -> '{s.replace('Python', 'World')}'")
print(f"[STR] .split()   -> {s.split()}")
print(f"[STR] original unchanged -> '{s}'")

word = "Python"
print(f"\n[SLICE] {word}[0:3]  -> '{word[0:3]}'")
print(f"[SLICE] {word}[::-1] -> '{word[::-1]}'")

print(f"\n[FMT] {salary:,.2f}")
print(f"[FMT] [{age:>5}]")
```

## 5. Conditionals
- `if / elif / else` checks conditions top to bottom and runs the **first** one that's true.
- **Ternary:** a one-line if, written `value_if_true if condition else value_if_false`.
- **match-case** (Python 3.10+): similar to `switch` in Java and C#. `case _` is the default.

```python
print("=" * 60)
print("SECTION 5: CONDITIONALS")
print("=" * 60)

def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 50:
        return "C"
    else:
        return "Fail"

for m in [95, 80, 60, 30]:
    print(f"[IF] marks={m:<3} -> grade {grade(m)}")

# Ternary
status = "adult" if age >= 18 else "minor"
print(f"\n[TERNARY] age {age} -> {status}")

# match-case
def http_status(code):
    match code:
        case 200: return "OK"
        case 404: return "Not Found"
        case 500: return "Server Error"
        case _:   return "Unknown"

for c in [200, 404, 999]:
    print(f"[MATCH] {c} -> {http_status(c)}")
```

## 6. Loops
- `for` loops over a sequence (list, string, range).
- `while` loops as long as a condition is true.
- `range(start, stop)` gives numbers from start up to **stop − 1**.
- `enumerate()` gives index **and** value together.
- `zip()` pairs items from two lists.
- `break` exits the loop. `continue` skips to the next iteration.
- **for-else:** the `else` block runs only if the loop **didn't** hit `break`.

```python
print("=" * 60)
print("SECTION 6: LOOPS")
print("=" * 60)

# for with range
print("[FOR] range(1, 6):", end=" ")
for i in range(1, 6):
    print(i, end=" ")
print()

# enumerate
fruits = ["apple", "banana", "mango"]
for idx, fruit in enumerate(fruits):
    print(f"[ENUMERATE] {idx} -> {fruit}")

# zip
prices = [100, 40, 60]
for fruit, price in zip(fruits, prices):
    print(f"[ZIP] {fruit} -> {price}")

# while
count = 3
while count > 0:
    print(f"[WHILE] count = {count}")
    count -= 1

# break and continue
for n in range(1, 10):
    if n % 2 == 0:
        continue
    if n > 7:
        print(f"[BREAK] stopped at {n}")
        break
    print(f"[CONTINUE] odd -> {n}")

# for-else
for n in [3, 5, 7]:
    if n % 2 == 0:
        break
else:
    print("[FOR-ELSE] loop finished without break")
```

## 7. Functions
- `def` defines a function. `return` sends a value back.
- **Default arguments:** a parameter with a preset value, like `greeting="Hello"`.
- **Keyword arguments:** pass by name, like `greet("Sita", greeting="Welcome")`.
- `*args` collects any number of positional arguments into a **tuple**.
- `**kwargs` collects any number of keyword arguments into a **dict**.
- `lambda`: a small one-line anonymous function.
- **Scope:** a variable assigned inside a function is **local**, so it doesn't change the global one.

```python
print("=" * 60)
print("SECTION 7: FUNCTIONS")
print("=" * 60)

# Default and keyword arguments
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print("[DEFAULT] ", greet("Ravi"))
print("[KEYWORD] ", greet("Sita", greeting="Welcome"))

# *args
def total(*args):
    print(f"   [ARGS] {args} -> type: {type(args).__name__}")
    return sum(args)

print(f"[ARGS] total = {total(1, 2, 3)}")

# **kwargs
def profile(**kwargs):
    print(f"   [KWARGS] {kwargs} -> type: {type(kwargs).__name__}")

profile(name="Venkatesh", role="Backend")

# lambda
square = lambda n: n * n
print(f"\n[LAMBDA] square(6) = {square(6)}")

# Scope
counter = 0
def increment():
    counter = 100
    print(f"   [SCOPE] inside function: counter = {counter}")

increment()
print(f"[SCOPE] outside function: counter = {counter}")
```
