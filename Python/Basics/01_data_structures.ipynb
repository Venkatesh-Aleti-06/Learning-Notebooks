## 0. Time and Space Complexity — Quick Reference

**Big-O** describes how the time or memory an operation needs grows as the data size (**n**) grows.

- **O(1)** — constant: same speed no matter the size (e.g. dict lookup)
- **O(log n)** — logarithmic: grows very slowly (e.g. binary search)
- **O(n)** — linear: grows with the size (e.g. searching a list)
- **O(n log n)** — e.g. sorting
- **O(n²)** — quadratic: nested loops; avoid on big data

### Time complexity by structure

| Operation | List | Tuple | Set | Dict |
|---|---|---|---|---|
| Access by index / key | O(1) | O(1) | — | O(1) |
| Search (`x in`) | O(n) | O(n) | **O(1)** | **O(1)** (keys) |
| Add at end | O(1)* | — | O(1) | O(1) |
| Insert / delete at start or middle | O(n) | — | — | — |
| Delete by value / key | O(n) | — | O(1) | O(1) |
| Sort | O(n log n) | — | — | — |

\* amortized: usually O(1), occasionally O(n) when the list resizes internally.
Set and dict values are **average case**; they use hashing.

### Space complexity
All four store **n** items, so space is **O(n)**.
Sets and dicts use **extra memory** for their hash tables, and tuples are slightly **lighter** than lists.

### Key interview point
To check whether an item exists, use a **set or dict (O(1))** instead of a list (O(n)). This one switch is behind many "optimize this" answers.

## 1. List
An **ordered**, **mutable** collection that allows **duplicates**. Written with `[ ]`.

| Method | What it does | Time |
|---|---|---|
| `append(x)` | add to end | O(1) |
| `extend(iter)` | add many to end | O(k) |
| `insert(i, x)` | add at position i | O(n) |
| `pop()` | remove last | O(1) |
| `pop(i)` | remove at position i | O(n) |
| `remove(x)` | remove first match by value | O(n) |
| `index(x)` | position of first match | O(n) |
| `count(x)` | how many times x appears | O(n) |
| `sort()` | sort in place | O(n log n) |
| `sorted(lst)` | return a new sorted list | O(n log n) |
| `reverse()` | reverse in place | O(n) |

**Trap:** `b = a` doesn't copy, both names point to the **same** list. Use `a.copy()` or `a[:]`.

```python
print("=" * 60)
print("SECTION 1: LIST")
print("=" * 60)

nums = [5, 2, 8, 2]
print(f"[CREATE]  {nums}")

# Access and slicing
print(f"[INDEX]   nums[0] = {nums[0]}, nums[-1] = {nums[-1]}")
print(f"[SLICE]   nums[1:3] = {nums[1:3]}")

# Adding
nums.append(10)
print(f"[APPEND]  {nums}")
nums.extend([1, 3])
print(f"[EXTEND]  {nums}")
nums.insert(0, 99)
print(f"[INSERT]  {nums}")

# Removing
last = nums.pop()
print(f"[POP]     removed {last} -> {nums}")
first = nums.pop(0)
print(f"[POP(0)]  removed {first} -> {nums}")
nums.remove(2)
print(f"[REMOVE]  removed first 2 -> {nums}")

# Searching
print(f"\n[INDEX]   position of 8 = {nums.index(8)}")
print(f"[COUNT]   count of 2 = {nums.count(2)}")
print(f"[IN]      8 in nums = {8 in nums}")
print(f"[LEN]     len = {len(nums)}")

# Sorting
print(f"\n[SORTED]  new list: {sorted(nums)}, original: {nums}")
nums.sort()
print(f"[SORT]    in place: {nums}")
nums.sort(reverse=True)
print(f"[DESC]    {nums}")

# Comprehension
squares = [n * n for n in range(1, 6)]
evens = [n for n in range(10) if n % 2 == 0]
print(f"\n[COMPREHENSION] squares = {squares}")
print(f"[COMPREHENSION] evens   = {evens}")

# Copy trap
a = [1, 2, 3]
b = a
c = a.copy()
a.append(4)
print(f"\n[ALIAS] a = {a}, b = {b}")
print(f"[COPY]  c = {c}")
```

## 2. Tuple
An **ordered**, **immutable** collection that allows **duplicates**. Written with `( )`.

- Once created, items **can't** be changed, added or removed.
- Only two methods: `count(x)` and `index(x)`, both O(n).
- **Unpacking:** `a, b = (1, 2)` assigns each item to a variable.
- **Single item needs a comma:** `(5,)` is a tuple, but `(5)` is just the number 5.
- Because it's immutable, a tuple can be a **dict key** or go in a **set**. A list can't.

**When to use:** fixed data that shouldn't change, like coordinates `(x, y)`, or returning several values from a function.

```python
print("=" * 60)
print("SECTION 2: TUPLE")
print("=" * 60)

point = (10, 20, 10)
print(f"[CREATE]  {point}")
print(f"[INDEX]   point[0] = {point[0]}, point[-1] = {point[-1]}")
print(f"[SLICE]   point[0:2] = {point[0:2]}")
print(f"[COUNT]   count of 10 = {point.count(10)}")
print(f"[INDEX]   position of 20 = {point.index(20)}")

# Immutability
try:
    point[0] = 99
except TypeError as e:
    print(f"\n[IMMUTABLE] {e}")

# Single item
print(f"\n[SINGLE] type of (5,) = {type((5,)).__name__}")
print(f"[SINGLE] type of (5)  = {type((5)).__name__}")

# Unpacking
x, y, z = point
print(f"\n[UNPACK] x={x}, y={y}, z={z}")

first, *rest = (1, 2, 3, 4)
print(f"[UNPACK] first={first}, rest={rest}")

# Returning multiple values
def min_max(values):
    return min(values), max(values)

low, high = min_max([4, 9, 1, 7])
print(f"[RETURN] low={low}, high={high}")

# Tuple as dict key
distances = {("Hyderabad", "Chennai"): 630}
print(f"\n[DICT KEY] {distances}")

try:
    bad = {[1, 2]: "list key"}
except TypeError as e:
    print(f"[DICT KEY] list fails -> {e}")
```

## 3. Set
An **unordered**, **mutable** collection of **unique** items. Written with `{ }`.

- Duplicates are removed automatically.
- No indexing, because there's no order.
- Items must be **hashable** (immutable), so numbers, strings and tuples are fine, but lists aren't.
- **Empty set:** use `set()`. `{}` creates an empty **dict**.

| Method / operator | What it does | Time |
|---|---|---|
| `add(x)` | add an item | O(1) |
| `remove(x)` | remove, error if missing | O(1) |
| `discard(x)` | remove, no error if missing | O(1) |
| `x in s` | membership check | **O(1)** |
| `a \| b` | union: everything in either | O(len a + len b) |
| `a & b` | intersection: common items | O(min(len a, len b)) |
| `a - b` | difference: in a but not b | O(len a) |
| `a ^ b` | symmetric difference: in one but not both | O(len a + len b) |

**When to use:** removing duplicates, and fast "does this exist?" checks.

```python
print("=" * 60)
print("SECTION 3: SET")
print("=" * 60)

s = {3, 1, 2, 3, 1}
print(f"[CREATE]  {s}")

print(f"[EMPTY]   type of set() = {type(set()).__name__}")
print(f"[EMPTY]   type of {{}}    = {type({}).__name__}")

# Remove duplicates from a list
emails = ["a@x.com", "b@x.com", "a@x.com"]
print(f"\n[DEDUP]   {emails} -> {set(emails)}")

# Add and remove
s.add(10)
print(f"\n[ADD]     {s}")
s.remove(10)
print(f"[REMOVE]  {s}")
s.discard(99)
print(f"[DISCARD] 99 not present, no error -> {s}")

try:
    s.remove(99)
except KeyError as e:
    print(f"[REMOVE]  missing item -> KeyError {e}")

# Membership
print(f"\n[IN]      2 in s = {2 in s}")

# Set operations
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(f"\n[SETS]    a = {a}, b = {b}")
print(f"[UNION]        a | b = {a | b}")
print(f"[INTERSECTION] a & b = {a & b}")
print(f"[DIFFERENCE]   a - b = {a - b}")
print(f"[SYMMETRIC]    a ^ b = {a ^ b}")

# Unhashable items
try:
    bad = {[1, 2]}
except TypeError as e:
    print(f"\n[HASHABLE] list in set -> {e}")
```

## 4. Dictionary
A collection of **key–value pairs**. **Mutable**, keys are **unique**, and since Python 3.7 it keeps **insertion order**. Written with `{key: value}`.

- Keys must be **hashable** (str, int, tuple). Values can be anything.
- `d[key]` raises `KeyError` if the key is missing. `d.get(key, default)` returns a default instead.

| Method | What it does | Time |
|---|---|---|
| `d[key] = v` | add or update | O(1) |
| `d.get(key, default)` | safe read | O(1) |
| `key in d` | check key exists | O(1) |
| `d.pop(key)` | remove and return value | O(1) |
| `del d[key]` | remove | O(1) |
| `d.update(other)` | merge another dict in | O(k) |
| `d.keys()` / `d.values()` / `d.items()` | views for looping | O(1) to create |
| `d.setdefault(key, v)` | read, or insert v if missing | O(1) |

**When to use:** fast lookup by a key, like a user by ID, counting things, or caching.
**In backend work:** JSON objects become Python dicts.

```python
# =====================================================================
# SECTION 4: DICTIONARY
# Paste this whole file into one Colab code cell.
# =====================================================================

# ---------- Create and read ----------
user = {"name": "Ravi", "age": 25, "role": "backend"}
print("created:", user)

print("name     ->", user["name"])
print("get city ->", user.get("city", "N/A"))

try:
    _ = user["city"]
except KeyError as e:
    print("KeyError:", e)

print("age in user ->", "age" in user)
print("len         ->", len(user))

# ---------- Add, update, delete ----------
user["city"] = "Hyderabad"
print("\nafter add city :", user)

user["age"] = 26
print("after update age:", user)

removed = user.pop("role")
print("popped role    :", removed, "->", user)

del user["city"]
print("after del city :", user)

# ---------- Merge ----------
user.update({"team": "backend", "age": 27})
print("\nafter update   :", user)

# ---------- Looping ----------
print("\nkeys  :", list(user.keys()))
print("values:", list(user.values()))
print("items :", list(user.items()))

for k, v in user.items():
    print(f"  {k} -> {v}")

# ---------- Count words with get ----------
words = ["red", "blue", "red", "green", "blue", "red"]
count = {}
for w in words:
    count[w] = count.get(w, 0) + 1
print("\nword count:", count)

# ---------- Dict comprehension ----------
squares = {n: n * n for n in range(1, 6)}
print("squares   :", squares)

prices = {"pen": 10, "book": 50, "bag": 400}
cheap = {k: v for k, v in prices.items() if v < 100}
print("cheap only:", cheap)
```

## 5. Bonus: collections module
- **Counter:** counts items in one line.
- **defaultdict:** a dict that creates a default value for missing keys, so no KeyError.
- **deque:** a double-ended queue. Adding or removing at **either end** is O(1), while a list is O(n) at the start. Used for BFS and sliding windows.

```python
from collections import Counter, defaultdict, deque

words = ["red", "blue", "red", "green", "red"]
c = Counter(words)
print("[COUNTER]", c)
print("[COUNTER] most common:", c.most_common(1))

groups = defaultdict(list)
for name, team in [("Ravi", "api"), ("Sita", "ai"), ("Anu", "api")]:
    groups[team].append(name)
print("\n[DEFAULTDICT]", dict(groups))

q = deque([1, 2, 3])
q.appendleft(0)
q.append(4)
print("\n[DEQUE]", q)
print("[DEQUE] popleft:", q.popleft(), "->", q)
```

## 6. Comparison: When to Use What

| | List | Tuple | Set | Dict |
|---|---|---|---|---|
| Syntax | `[1, 2]` | `(1, 2)` | `{1, 2}` | `{"a": 1}` |
| Ordered | Yes | Yes | No | Yes (3.7+) |
| Mutable | Yes | No | Yes | Yes |
| Duplicates | Yes | Yes | No | Keys: no |
| Search | O(n) | O(n) | O(1) | O(1) |

- **List:** ordered data you'll change, like a queue of tasks.
- **Tuple:** fixed data, like coordinates or multiple return values.
- **Set:** unique items and fast existence checks.
- **Dict:** lookup by key, like users by ID, counts, or caching.

**Interview line:** "I pick the structure by the operation I need most. For fast lookups I use a set or dict, which is O(1), instead of a list, which is O(n)."
